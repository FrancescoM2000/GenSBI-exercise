```python
u = np.linspace(-8,8,16)
v = np.linspace(-8,8,16)

print(u)
print(v)

grid = torch.zeros(16,16,2)
idx = np.linspace(0,15,15+1)
idx = np.asarray(idx, dtype="int")

print(idx)

for i in idx:
    for j in idx:
        grid[i][j][0] = u[i]
        grid[i][j][1] = v[j]

print(grid/8.)

```

```python
Nside = 16
u = torch.linspace(-1., 1., Nside)
v = torch.linspace(-1., 1., Nside)

U, V = torch.meshgrid(u, v, indexing="ij")

coords = torch.stack([U, V], dim=-1)
```

```python
# Pixel boundaries
edges = torch.linspace(-1., 1., Nside + 1)

# Pixel centers
u = (edges[:-1] + edges[1:]) / 2
v = u.clone()

# Spatial grid
U, V = torch.meshgrid(u, v, indexing="ij")

# Coordinate features
grid = torch.stack([U, V], dim=-1)
```
